```python
import numpy as np
import matplotlib.pyplot as plt


#### Import Mean Temp Data ###

from data_analysis_script import Dataset
run_label = '2024-05-30' ## Change this to look at different runs
d = Dataset(run_label=run_label, gases=['O2'], data_path='20240530/')
d.GetData()

t = np.array(d.data['O2']['Exposure_time'])             
temp = np.array(d.data['O2']['Mean_temp']) 


### Create Decay Function ###

def exponential_decay(t, tau):
    return np.exp(-t/tau) * (t >= 0)

### Convolve Mean Temp Data with Decay Function into 'Smooth_Step_Function' ###

def smooth_step_function(t, tau):
#     step = temp
    dt = t[1] - t[0]  # Assuming t is evenly spaced
    decay_t = np.arange(0, len(t)) * dt  # Define time for the decay function
    decay = exponential_decay(decay_t, tau)
    Normdecay = decay/np.sum(decay) #Normalize decay function

    padded_data = np.pad(temp, (len(decay)-1, 0), mode='edge')  # Pad data with values similar to edge to resolve boundary issues
    smooth_step = np.convolve(padded_data, Normdecay, mode='valid') #Convolve two functions
    return smooth_step


### Plot the results for different tau values ###

taus = [1000, 2500, 5000, 7500, 10000]  # smaller tau values for tighter fit
plt.figure(figsize=(10, 6))

for tau in taus:
    smoothed = smooth_step_function(t, tau)   
    plt.plot(t, smoothed, label=f'Smoothed Step Function (tau={tau})')
  
    
    
plt.plot(t, temp, 'k--', label='Mean temp data')
plt.xlabel('Time [s]')
plt.ylabel('Temp [K]')
plt.ylim(292, 325)
plt.legend()
plt.title('Step Function Smoothing with Different Tau Values')
plt.grid()
plt.savefig("graph")
plt.show()

```
